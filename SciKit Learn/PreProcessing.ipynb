# Preprocessing is the process of preparing the data for the model to use, by sorting, classifying and removing corrupt and redundant parts


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

#Standardize features by removing the mean and scaling to unit variance.
from sklearn.preprocessing import StandardScaler
```

```python
X,Y = load_iris(return_X_y= True)
X

#The features cam have a wide range of values, these differences in ranges creates a problem for algorithms like Kneighbours.
#As KNeighbours rely on the classification of most nearest neighbours to classify data.
#Class one is usually in the range of 3-5.5, amd the third between 0-1, without preprocessing, one would dominate over three in KNN
```

```python
#Splitting data in training and testing
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size= 0.2)
```

```python
scaler = StandardScaler()

'''
fit_transform(X, y=None, **fit_params)
Fit to data, then transform it.

Fits transformer to X and y with optional parameters fit_params and returns a transformed version of X.
'''

'''
fit(X, y=None, sample_weight=None)[source]
Compute the mean and std to be used for later scaling.
'''
X_train_scaled = scaler.fit_transform(X_train)


X_test_scaled = scaler.transform(X_train)

X_train_scaled
```
