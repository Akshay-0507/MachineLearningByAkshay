# Feature Encoding:
 the process of converting non-numerical data (like text or categories) into numerical values.

1.OneHotEncoder - Used for Nominal data (no inherent order like Color, City) 

2.OrdinalEncoder - Ordinal features (ordered categories like Size: S, M, L) 

3.TargetEncoder - High-cardinality nominal data (e.g., thousands of zip codes) 

4.LabelEncoder - Target labels (y) only, not input features (X) 



```python
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import OrdinalEncoder
```

```python

data = fetch_openml('car', as_frame= True).frame
data
```

```python
# Method 1: Specify categories manually (recommended for ordinal features)
columns_to_encode = ['lug_boot', 'safety']

encoder = OrdinalEncoder()
data[columns_to_encode] = encoder.fit_transform(data[columns_to_encode])

print("After encoding:")
data

#Encoder can also be reversed
```
