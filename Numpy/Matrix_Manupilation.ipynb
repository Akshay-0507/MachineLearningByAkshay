```python
arr = np.array([[1,2],[4,5]])
arr
```

```python
#TRANSPOSE:
print(arr.transpose())
```

```python
arr2 = np.array([[[1,2],[4,5]]])
print(arr2.shape)
swap = np.swapaxes(arr2, 0, 1)
print(swap.shape)
print(swap)
```
