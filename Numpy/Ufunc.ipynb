A universal function (or ufunc for short) is a function that operates on ndarrays in an element-by-element fashion, supporting array broadcasting, type casting, and several other standard features. That is, a ufunc is a “vectorized” wrapper for a function that takes a fixed number of specific inputs and produces a fixed number of specific outputs


```python
import numpy as np
```

```python
arr1 = np.array([1,4,9,16])
arr1
```

```python
#SQRT FUNC
print(np.sqrt(arr1))
```

```python
#EXPONENTION:
print(np.exp(arr1))# return e^x, where e is exponential and x is the element
```

```python
#SINE function: np.sine()
arr2 = np.array([0,np.pi, np.pi/2])
print(np.sin(arr2))
```

```python
#TO iterate through a numpy array we use np.iter()
arr3 = np.array([[1,2],[4,5]])
for x in np.nditer(arr3):
    print(x, end = " ")

#ndenumerate -> index + values.
```
