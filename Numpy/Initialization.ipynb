```python
import numpy as np
```

#There are multiple ways to initialize a array


```python
#FIRST METHOD
#np.arange(start,end,jump = 1)
ar = np.arange(0,10)
ar
```

```python
#SECOND METHOD
#Allows to create even spaced data seet,
#np.linspace(start,end, seperation between values)
ar2 = np.linspace(0,1,5)
ar2
```

```python
#THIRD METHOD
#np.logspace(start,end,points)
ar3 = np.logspace(1,3,4) #10^1 -> 10^3 with four points in between
ar3
```

```python
#Create a array full of zeros:
ar4 = np.zeros([2,3]) #array of zeros in two rows and three columns
print(ar4)
print("--------------------------------------------------------\n")
#Array of ones:
ar5 = np.ones([4,2],dtype = float)
print(ar5)
print("--------------------------------------------------------\n")

#Create a array full of any val
#np.full(number of values, value)
np.full(10,2)
```

```python
#Random numbers with uniform distribution
np.random.rand(2,3) # returns float values between 0->1
```

```python
#Generate random numbers with normal distribution, ie mean = 0 sd = 1
np.random.randn(2,3)
```

```python
#To produce array of random integers within an range:
np.random.randint(1,10, (2,3)) #start,stop,dimension
```
