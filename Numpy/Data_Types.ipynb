```python
import numpy as np
```

NUMPY data types can have many data types. But each single array can only store one type of data.
Each data type have a priority. point values have higher priority than integer type.
If there is a decimal value in the list, then all the integer values in the list will be converted to the decimal types.


```python
lst = ["string","poass",1,2,3]
npar = np.array(lst)#all the non-string values will be implicitly coverted to string.
npar
#String > Float > int
```

```python
#Explicit type casting
arr1 = np.array([1,2,3,4,5], dtype = np.float64) #By default the values are int64
print(arr1, "\n", arr1.dtype)
#The inverse is also true, float values can also be explicitly be converted to int
```

TYPECASTING:
Done using the keyword -> astype()
It creates the copy of the orignal array,no changes are done to the orignal.


```python
arr2 = np.array([1,2,3])
#converting the array from int64 to float64
arr2fl = arr2.astype(np.float64)
arr2fl
#WE cannot convert characters to int or float
```
