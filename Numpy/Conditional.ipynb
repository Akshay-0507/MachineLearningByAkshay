```python
import numpy as np
```

```python
#WHERE
arr = np.array([1,2,3,4,5,6,7])
print(arr)
```

```python
result = np.where(arr < 2, "low", "high")#NUMPY.Where((CONDITION), TRUE result, False result)
print(result)
```

```python
#ARGWHERE : returns the row or column position where the argument is true.
arr2 = np.array([10,20,30,40,50])
print(np.argwhere(arr2 > 20))
```

```python
#LOGICAL AND: returns a array of False and True values.
mask = np.logical_and(arr2 > 15, arr2<45)
print(mask)
```
