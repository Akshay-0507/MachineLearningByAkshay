WE can perform mathematical operations on the entire elements without having to loop through the array


```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
```

```python
#ADDITION:
print(a + b)
#Each element will be added to its corresponding element
```

```python
#SUBTRACTION:
print(b-a)
```

```python
#DIVISION:
print(a/b)
print(b//a)
```

```python
#MULTIPLICATION:
print(a*b)
```

```python
#Modulus:
print(b%a)
```

```python
#EVERY element raised to 2
print(a**2)
```
