View = changes are reflected upon the orignal array

copy =  changes are only applied to the new array or the copy array

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6])
view = arr[1:3]
view
```

```python
view[0] = 200#this will also change the values in the orignal array
print(view)
print(arr)
```

```python
copy = arr[1:3].copy()
copy[0] = 2#this will not change the orignal arr
print(copy)
print(arr)
```
