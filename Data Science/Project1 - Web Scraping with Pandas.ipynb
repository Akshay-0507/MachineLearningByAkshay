### 2. Reading from multiple .csv files
    
       Link Structure: Root + season + league

https://www.football-data.co.uk/mmz4281/2122/E0.csv
\
https://www.football-data.co.uk/mmz4281/2122/E1.csv
\
https://www.football-data.co.uk/mmz4281/2122/E2.csv
\
https://www.football-data.co.uk/mmz4281/2122/E3.csv
\
https://www.football-data.co.uk/mmz4281/2122/EC.csv

```python
#Create a root variable
root = "https://www.football-data.co.uk/mmz4281/"
```

```python
#Creating a list of leagues:
leagues = ['E0', 'E1','E2','E3','EC']

#Creating a list of dataframes
frames = []

#Looping through leagues, read multiple csv and append it into a list:

for league in leagues:
    #Without specifying the encoding it will throw error: UnicodeDecodeError
    #encoding='unicode_escape' is a special codec used to convert Unicode strings into a byte sequence (or vice versa) 
    #that represents characters using Python's string literal escape sequences (e.g., \uXXXX or \xXX)
    df = pd.read_csv(root + '2122'+ '/' + league + '.csv', encoding='unicode_escape')
    frames.append(df)
```

```python
#Checking for length of Frames:
len(frames)
```

```python
#Show the 1st element
frames[0]
```


## 2.2 Multiple seasons

```python
#Looping through range of seasons and inserting new columns:
#Creating a list of leagues:
leagues = ['E0','E2','E3']

#Creating a list of dataframes
frames = []

#Looping through range of seasons and inserting new columns:
for league in leagues:
    for season in range(15, 21):
        df = pd.read_csv(root + str(season) + str(season +1)+ '/' + league + '.csv', encoding='unicode_escape')
        #df.insert(index position, name of column, values)
        df.insert(1, 'season', season)
        frames.append(df)
    
```

```python
#Total frames = season * league = 18
len(frames)
```

```python
#Showing 1st and last column
frames[-1]
```
