# Identifying the missing data


```python
#number of rows missing in each column
df_netflix.isnull().sum().sort_values(ascending = False)
```

```python
# %of rows missing in each column out of total
for column in df_netflix.columns:
    print(column, ":", df_netflix[column].isnull().mean() * 100)
```

# 3. Dealing with missing data

```python
#Remove a column or row with .drop, or .dropna, .isnull
no_director = df_netflix[df_netflix["director"].isnull()].index #Gives the list of indexes with null values in the director column

df_netflix.drop(no_director, axis = 0)
```

## Filling the missing data


```python
#Use fillna() to fill NaN with mean or median or mode or any other value
mode = ''.join(df_netflix["rating"].mode())

df_netflix['rating'].fillna(mode, inplace = True)
```

```python
#Extracting data
df_netflix['duration']
```

```python
#Filtering out TV shows
df_movie = df_netflix[df_netflix['type'] == 'Movie']
df_movie['duration']
```
