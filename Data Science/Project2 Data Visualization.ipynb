```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Importing .csv file
df_pop = pd.read_csv("https://raw.githubusercontent.com/thepycoach/python-for-data-science/main/07.Project%202%20-%20Data%20Visualization/population_total.csv")
df_pop
```

### 1.Making a Pivot Table

```python
#Dropping null values
df_pop.dropna(inplace = True)
```

```python
#Pivot table
df_pop_pivot = df_pop.pivot(index = "year", columns = "country", values = "population")
```

```python
#Selecting some country:
df_pop_pivot = df_pop_pivot[['United States', 'India', 'China', 'Indonesia', 'Brazil']]
```

# 2.Lineplot


```python

df_pop_pivot.plot(kind = 'line', xlabel = 'Year', ylabel= 'Population', title = 'PopulationGrowth', figsize = (8,4))

#Saving the plot 
plt.savefig("my_test.png")
#Showing the figure
plt.show()
```

# 3.BarPlot

### 3.1 Single BarPlot

```python
#selecting only single year
df_pivot_2020 = df_pop_pivot[df_pop_pivot.index.isin([2020])]

#IN BARGRAPH THE INDEX NEEDS TO BE IN STRING:
df_pivot_2020t = df_pivot_2020.transpose()
```

```python
#Making barplot
df_pivot_2020.plot(kind = 'bar')
```

## 3.2 Barplot Group by n variables:


```python
df_pivot_decade = df_pop_pivot[df_pop_pivot.index.isin([1980,1990,2000,2010,2020])]
df_pivot_decade.plot(kind ="bar")
```

```python
df_pivot_2020t.plot(kind = "pie", y ='2020' )
```
