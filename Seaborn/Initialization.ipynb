```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
penguin = sns.load_dataset('penguins')
```

```python
penguin.head()
```

```python
penguin['species'].value_counts()
```

```python
penguin['island'].value_counts()
```

```python
#Making a scatter plot
sns.scatterplot(data = penguin, x = "flipper_length_mm", y = "body_mass_g", hue = "sex")
```

```python
sns.scatterplot(data = penguin, x = "species", y = "body_mass_g", hue = "island", style ="sex", alpha = 0.5 )
```

```python
#Swarm plot
sns.swarmplot(data = penguin, x = "species", y = "body_mass_g", hue = "island")
```
