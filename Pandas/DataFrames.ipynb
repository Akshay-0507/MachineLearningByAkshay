```python
import pandas as pd
import numpy as np
```

```python
Data = {
    "Name": ["Alex","Bob","Charlie","David","Eve","Alex"],
    "Age" : [25,30,35,np.nan,29,25],
    "Department" : ["HR","IT","Finance","IT","HR","HR"],
    "Salary" : [50000,60000,70000,62000,np.nan,50000]
}
```

```python
Data
```

```python
#Converting the above data into DataFrame:
df = pd.DataFrame(Data)
df
```

```python
#To see only the first two rows:
print(df.head(2))
print("-----------------------------------------------------\n")

#To see the last two values:
print(df.tail(2))
```

```python
#Using iloc with df
df.iloc[1:3] #iloc[Starting row : Ending Row(Not included), :Columns to included(All by default)]
```

```python
#Filetering using loc with departments:
df.loc[1:3, ["Name","Age","Department"]]#loc[Starting row : Ending Row, :Columns to included(All by default,Using column name)]
```

```python
#Droping a column:
df.drop("Age", axis = 1)#axi =1 drops the entire column
#The column age is not deleted rather it is hidden for this output
#To make the changes permanent we use the keyword "inplace = True"
```

```python
#To check Info of dataFrame:
print(df.info())
```

```python
#To get statistical Information about the dataset we use the describe function
df.describe()
```

```python
#DataFrames also support broadcasting:
df["Salary"] = df["Salary"] + 5000
df
```

```python
#Renaming columns:
df.rename(columns = {"Department" : "Dept"}, inplace = True)
```

```python
#To find the number of unique elements in a column:
print(df["dept"].unique())

#To find the frequency of each unique elements:
print(df["dept"].value_counts())
```
