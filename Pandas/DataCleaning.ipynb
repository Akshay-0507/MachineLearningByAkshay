```python
import pandas as pd
import numpy as np
```

```python
Data = {
    "Name": ["Alex","Bob","Charlie","David","Eve","Alex"],
    "Age" : [25,30,35,np.nan,29,25],
    "Department" : ["HR","IT","Finance","IT","HR","HR"],
    "Salary" : [50000,60000,70000,62000,np.nan,50000],
    "Promoted_Salary" : [52000,62000,72000,64000,np.nan,52000]
}
```

```python
df = pd.DataFrame(Data)
df
```

```python
#Checking how many nulls values there are in the dataframe:
df.isnull().sum()
#The data represents how many null values each column has.
```

```python
#To drop rows containing null values:
print(df.dropna())

print("\n------------------------------------------------------------------\n")
#To drop the row only if all the data in the row is null:
print(df.dropna(how = "all"),)
print("\n------------------------------------------------------------------\n")
```

Removing null values by filling them using fillna method:

```python
#Filling with zero:
df.fillna(0)
print(df)
```

```python
#Filling the null values with the mean of those columns:
df["Age"].fillna(df["Age"].mean())
```

```python
#Filling the null value with median:
df["Salary"].fillna(df["Salary"].median())
```

There are two different methods of filling null values without specifying data
FowardFill"ffill" and BackwardFill "bfill"
FowardFill Fills the null value with the last value encountered,
Reverse for the Backward fill


```python
#Chanig the values in the clumns:
df["Name"] = df["Name"].replace("Charlie", "Rose")
```

```python
df
```

REMOVING THE DUPLICATE ELEMENTS:


```python
#Finding the duplicates:
df_dup = df[df.duplicated(keep = "first")]
df_dup
```

```python
#To drop the duplicate value
df = df.drop_duplicates()
df
```

```python
#apply and lambda helps change single entries based on certain conditions.
# def multiply_age(x):
#     return x*2
# df["Age"] = df["Age"].apply()
df["Age"] = df["Age"].apply(lambda x: x/2)
df
```

JOIN AND MERGE:

```python
department_info = {
    "Department" : ["HR", "IT", "Finance"],
    "Location" : ["New York", "San Francisco", "Chicago"],
    "Manager" : ["Laura", "Steve", "Nina"]
}
df2 = pd.DataFrame(department_info)
df2
```

```python
#Concat: Horizontal stacking
pd.concat([df,df2])
```

```python
#Concat based on department:
pd.concat([df,df2], axis = 1)
```

```python
#Merging withgout repreating columns:
pd.merge(df,df2, on = "Department")
```
