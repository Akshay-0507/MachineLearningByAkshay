```python
import pandas as pd
```

```python
sr = pd.Series([1,2,3,4,5,6,7,8,9])
sr
```

```python
#Checking the data type of series:
print(sr.dtype, "\n")

#Checking the values in the series:
print(sr.values,"\n")

#Checking the indexes of the series:
print(sr.index, "\n")
```

```python
#Assining the name to the column:
sr.name = "Numbers"
sr.name
```

Indexing in the PANDAS is same as in normal python list or NUMPY arrays.

```python
#Another way of indexeing in pandas is 
#iloc 
print(sr.iloc[3])

#Fetching multiple indexes at once
print(sr.iloc[[1,2,4,5]])
```

```python
#Chaning the default indexing
index = ["Apple", "Banana","Lichi","Mango","Berry","Pineapple","Jackfruit","Pine","Shrew"]
sr.name = "Calories"
sr.index = index
sr
```

```python
#Now the new indexes are the location addresses, now iloc(1) will not work as index 1 do not exists:
#print(sr.iloc(1)) -> THROWS ERROR

#Now to retrieve values using the new indexes we use another function called loc()
print(sr.loc["Apple"])
#In loc or location based indexeing the stop values, or the end indexing is also printed
```

```python
#We can also create Series using Dictionary:
# An example dictionary
data = {'Russia': 'Moscow', 'Hungary': 'Budapest', 'Switzerland': 'Bern'}
sr2 = pd.Series(data)
sr2.name = "Capital"
print(sr2)
```
